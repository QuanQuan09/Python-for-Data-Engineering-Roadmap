```python
# Task 1 - Load all 3 CSVs

import pandas as pd

servers = pd.read_csv("servers.csv")
metrics = pd.read_csv("metrics.csv")
owners = pd.read_csv("owners.csv")

print(servers)
print(metrics)
print(owners)
```

```python
# Task 2 - Merge servers with metrics

import pandas as pd

server_report = pd.merge(servers, metrics, on="server_id")

print(server_report)
```

```python
# Task 3 - Merge server_report with owners

import pandas as pd

server_report = pd.merge(server_report, owners, on="server_id")

print(server_report)
```

```python
# Task 4 - Display only servers that are down

import pandas as pd 

down_servers = server_report[server_report["status"] == "Down"]

print(down_servers)
```

```python
# Task 5 - Display only servers with cpu_usage > 80

import pandas as pd

server_report[server_report["cpu_usage"] > 80]
```

```python
# Task 6 - Sort by cpu_usage

print(server_report.sort_values("cpu_usage"))
```

```python
# Task 7 - Group By environment

server_report.groupby("environment")["cpu_usage"].mean()
```

```python
# Task 8 - Create a new column named health
import pandas as pd

server_report["health"] = "Healthy"

server_report.loc[server_report["cpu_usage"] >= 80, "health"] = "Warning"

server_report.loc[server_report["cpu_usage"] >= 90, "health"] = "Critical"

print(server_report)
```

```python
# Export final report

server_report.to_csv("server_health_report.csv", index=False)
```

```python
# Bonus Challenge - Create a health summary report

print(server_report.groupby("health")["cpu_usage"].agg(["count", "sum", "mean", "max"]))
```

```python
# Bonus Challenge - Create health summary report

total_servers = len(server_report)
servers_up = len(server_report[server_report["status"] == "Up"])
servers_down = len(server_report[server_report["status"] == "Down"])
average_cpu = server_report["cpu_usage"].mean()
critical_servers = len(server_report[server_report["health"] == "Critical"])
warning_servers = len(server_report[server_report["health"] == "Warning"])
healthy_servers = len(server_report[server_report["health"] == "Healthy"])

print("==== SERVER HEALTH REPORT ====")
print()
print(f"Total Servers: {total_servers}")
print(f"Servers Up: {servers_up}")
print(f"Average CPU: {average_cpu}")
print(f"Critical Servers: {critical_servers}")
print(f"Warning Servers: {warning_servers}")
print(f"Healthy Servers: {healthy_servers}")
```

```python
import pandas as pd

data = {
    "server_id": [1, 2, 3, 4],
    "cpu_usage": [45, 85, 65, 91],
    "status": ["Up", "Down", "Up", "Down"]
}

df = pd.DataFrame(data)

df.to_csv("metrics.csv", index=False)
```

```python
# Lesson 1 - Load Multiple CSV Files

import pandas as pd

servers = pd.read_csv("servers.csv")
metrics = pd.read_csv("metrics.csv")

print(servers)
print(metrics)
```

```python
# Lesson 1a - Merge Two DataFrames
# works like a SQL join - servers has server details, metrics has monitoring details
# merge() combines them using the matching server_id

import pandas as pd

merged_df = pd.merge(servers, metrics, on="server_id")

print(merged_df)
```

```python
merged_df = pd.merge(servers, metrics, on="server_id", how="inner")
```

```python
print(merged_df)
```

```python
day1 = pd.DataFrame({
    "server": ["WEB01", "WEB02"],
    "cpu": [45, 78]
})
```

```python
day2 = pd.DataFrame({
    "server": ["WEB03", "DB01"],
    "cpu": [65, 52]
})
```

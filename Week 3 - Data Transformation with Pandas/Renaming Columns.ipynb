```python
import pandas as pd

data = {
    "Server Name": ["WEB01", "WEB02", "DB01"],
    "CPU Usage (%)": [55, 92, 70],
    "Status": ["Up", "Down", "Up"]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Rename columns to create a cleaner DataFrame

df = df.rename(columns={
    "Server Name": "server",
    "CPU Usage (%)": "cpu_usage",
    "Status ": "status"
})

print(df)
```

```python
# Run code that comes from a messy source

import pandas as pd

data = {
    "Device Name": ["Router01", "Switch01", "Firewall01"],
    "Device Status ": ["Online", "Offline", "Online"],
    "Memory Usage (%)": [65, 88, 72]
}

df = pd.DataFrame(data)

df = df.rename(columns={
    "Device Name": "device",
    "Device Status ": "status",
    "Memory Usage (%)": "memory_usage"
})

print(df)
```

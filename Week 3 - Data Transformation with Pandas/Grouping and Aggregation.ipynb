```python
# Summaries are more important that individual rows

import pandas as pd

data = {
    "environment": ["Prod", "Prod", "Dev", "Dev", "Prod"],
    "cpu_usage": [80, 90, 50, 60, 70]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Average CPU by Environment

df.groupby("environment")["cpu_usage"].mean()
```

```python
# Counting Records

df.groupby("environment").size()
```

```python
# Sum Values

df.groupby("environment")["cpu_usage"].sum()
```

```python
import pandas as pd

data = {
    "status": ["Up", "Down", "Up", "Down", "Up"],
    "memory_usage": [45, 88, 72, 95, 60]
}

df = pd.DataFrame(data)

print(df.groupby("status")["memory_usage"].mean())
```

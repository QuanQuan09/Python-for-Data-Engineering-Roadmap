```python
# Creating New Information from Existing Data
# Server CPU Utilization

import pandas as pd

data = {
    "server": ["WEB01", "WEB02", "DB01"],
    "cpu_usage": [55, 92, 70]
}

df = pd.DataFrame(data)

df["cpu_remaining"] = 100 - df["cpu_usage"]

print(df)
```

```python
# Create a warning flag for heavily loaded servers

df["high_cpu"] = df["cpu_usage"] > 80

print(df)
```

```python
import pandas as pd

data = {
    "server": ["WEB01", "WEB02", "WEB03"],
    "memory_usage": [45, 88, 67]
}

df = pd.DataFrame(data)

df["memory_remaining"] = 100 - df["memory_usage"]

df["high_memory"] = df["memory_usage"] > 75

print(df)
```

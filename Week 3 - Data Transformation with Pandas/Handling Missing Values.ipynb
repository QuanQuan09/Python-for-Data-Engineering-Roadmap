Real-world datasets almost always have missing data

```python
import pandas as pd

data = {
    "server": ["WEB01", "WEB02", "DB01", "WEB03"],
    "cpu_usage": [55, None, 70, None]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Find missing values

print(df.isnull())
```

```python
# Counting Missing Values

print(df.isnull().sum())
```

```python
# Filling Missing Values
# Replace missing CPU values with 0

df["cpu_usage"] = df["cpu_usage"].fillna(0)

print(df)
```

```python
# Replace missing values with the average

avg_cpu = df["cpu_usage"].mean()

df["cpu_usage"] = df["cpu_usage"].fillna(avg_cpu)

print(df)
```

```python
import pandas as pd

data = {
    "server": ["WEB01", "WEB02", "WEB03", "DB01"],
    "memory_usage": [45, None, 72, None]
}

df = pd.DataFrame(data)

print(df.isnull().sum())

df["memory_usage"] = df["memory_usage"].fillna(0)

print(df)
```

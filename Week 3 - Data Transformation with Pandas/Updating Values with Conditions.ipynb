# Using .loc

```python
# .loc
# Select Rows

import pandas as pd

data = {
    "server": ["WEB01", "WEB02", "DB01"],
    "cpu_usage": [55, 92, 70]
}

df = pd.DataFrame(data)
```

```python
# All rows where CPU is greater than 80

df.loc[df["cpu_usage"] > 80]
```

```python
# Select Specific Columns

df.loc[df["cpu_usage"] > 80, "server"] # locate all rows where cpu_usage is greater than 80 & only show server column
```

# Begin Lesson 3 - Updating Values with Conditions  

Data Engineers frequently create categories based on conditions  
Instead of storing:  
92, 70, 45  

We might want:  
Critical, Warning, Normal

```python
import pandas as pd

data = {
    "server": ["WEB01", "WEB02", "DB01"],
    "cpu_usage": [55, 92, 70]
}

df = pd.DataFrame(data)

df["cpu_status"] = "Normal"

df.loc[df["cpu_usage"] > 80, "cpu_status"] = "Critical"

print(df)
```

```python
# Multiple Conditions

df["cpu_status"] = "Normal"

df.loc[df["cpu_usage"] >= 70, "cpu_status"] = "Warning"

df.loc[df["cpu_usage"] >= 90, "cpu_status"] = "Critical"

print(df)
```

```python
import pandas as pd

data = {
    "server": ["WEB01", "WEB02", "WEB03", "DB01"],
    "memory_usage": [45, 88, 72, 95]
}

df = pd.DataFrame(data)

df["memory_status"] = "Normal"

df.loc[df["memory_usage"] >= 70, "memory_status"] = "Warning"

df.loc[df["memory_usage"] >= 90, "memory_status"] = "Critical"

print(df)
```

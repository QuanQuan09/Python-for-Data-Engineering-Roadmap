After cleaning data, we often want to sort it

```python
import pandas as pd

data = {
    "server": ["WEB01", "WEB02", "DB01", "WEB03"],
    "cpu_usage": [55, 92, 70, 45]
}

df = pd.DataFrame(data)

print(df.sort_values("cpu_usage"))
```

```python
# Descending Order

df.sort_values("cpu_usage", ascending=False)
```

```python
# Sorting by Multiple Columns

df.sort_values(
    ["server", "cpu_usage"],
    ascending=[True, False]
)
```

```python
import pandas as pd

data = {
    "server": ["WEB01", "WEB02", "WEB03", "DB01"],
    "memory_usage": [45, 88, 72, 95]
}

df = pd.DataFrame(data)

sorted_df = df.sort_values(
    "memory_usage",
    ascending=False
)

print(sorted_df)
```

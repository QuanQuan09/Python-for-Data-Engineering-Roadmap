```python
import pandas as pd

data = {
    "server": [
        "web01 ",
        " WEB02",
        "db01",
        "WEB03"
    ],
    "status": [
        "Up",
        "Down",
        "Up",
        "Down"
    ],
    "cpu_usage": [
        "55",
        "92",
        None,
        "70"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Task 1 - Clean the server column so all values:
# Have no extra spaces
# Are uppercase

df["server"] = df["server"].str.strip()
df["server"] = df["server"].str.upper()

print(df)
```

```python
# Task 2 - Replace missing cpu_usage values with 0

df["cpu_usage"] = df["cpu_usage"].fillna(0)
print(df)
```

```python
# Task 3 - convert cpu_usage from strings to integers

df["cpu_usage"] = df["cpu_usage"].astype(int)
print(df.dtypes)
```

```python
# Task 4 - Create a new cpu_status column where:
# cpu_usage >= 90 -> Critical
# cpu_usage > 70 -> Warning

df["cpu_status"] = "Normal"

df.loc[df["cpu_usage"] >= 70, "cpu_status"] = "Warning"

df.loc[df["cpu_usage"] >= 90, "cpu_status"] = "Critical"

print(df)
```

```python
# Task 5 - Sort by cpu_usage descending

df.sort_values("cpu_usage", ascending=False)
```

```python
# Task 6 - Calculate:
# Average cpu_usage by status using groupby()

df.groupby("status")["cpu_usage"].mean()
```

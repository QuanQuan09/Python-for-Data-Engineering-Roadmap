Converting data from one type to another

```python
# Converting Strings to Integers

import pandas as pd

data = {
    "cpu_usage": ["85"]
}

df = pd.DataFrame(data)

df["cpu_usage"] = df["cpu_usage"].astype(int)

print(df.dtypes)
```

```python
import pandas as pd

data = {
    "server": ["WEB01", "WEB02", "WEB03"],
    "cpu_usage": ["55", "92", "70"]
}

df = pd.DataFrame(data)

print(df.dtypes)

df["cpu_usage"] = df["cpu_usage"].astype(int)

print(df.dtypes)
```

```python
import pandas as pd

data = {
    "cpu_usage": ["55", "92", "Unknown"]
}

df = pd.DataFrame(data)

pd.to_numeric(df["cpu_usage"], errors="coerce")

print(df)
```

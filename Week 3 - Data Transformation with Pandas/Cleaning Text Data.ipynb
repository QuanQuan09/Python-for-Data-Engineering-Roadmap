One of the most common real-world data problems is inconsistent text  
Problems that exist:  
- Extra spaces   
- Mixed capitalization
- Inconsistent formatting

These issues can break reports, joins, filters, and database loads

```python
import pandas as pd

server_data = {
    "server": [
        "web01",
        " WEB02 ",
        "Db01",
        "WEB03 "
    ]
}

df = pd.DataFrame(server_data)

# Removing Extra Spaces

df["server"] = df["server"].str.strip()
print(df)
```

```python
# Removing Extra Spaces
# Changing server data to upper case

import pandas as pd

data = {
    "server": [
        "web01",
        " WEB02 ",
        "Db01",
        "WEB03 "
    ]
}

df = pd.DataFrame(data)

df["server"] = df["server"].str.strip().str.upper()

print(df)
```

```python
# Prepare and Clean the Data

df["incident_time"] = pd.to_datetime(
    df["incident_time"]
)

df["resolved_time"] = pd.to_datetime(
    df["resolved_time"]
)

df["incident_date"] = (
    df["incident_time"].dt.date
)

df["incident_weekday"] = (
    df["incident_time"].dt.day_name()
)

df["resolution_minutes"] = (
    (
        df["resolved_time"]
        - df["incident_time"]
    )
    .dt.total_seconds()
    / 60
)

print(
    df[
        [
            "incident_id",
            "server",
            "team",
            "environment",
            "status",
            "priority",
            "incident_date",
            "resolution_minutes"
        ]
    ]
)
```

```python
# Build a Team Performance Report
# Create a report grouped by team

team_report = (
    df.groupby("team")
      .agg(
          incident_count=("incident_id", "count"),
          average_cpu_usage=("cpu_usage", "mean"),
          highest_cpu_usage=("cpu_usage", "max"),
          average_memory_usage=("memory_usage", "mean"),
          average_resolution_minutes=(
              "resolution_minutes",
              "mean"
          )
      )
      .reset_index()
      .round(2)
)

print(team_report)
```

```python
# Sort Report from Largest to Smallest Incident Count

team_report = team_report.sort_values(
    by="incident_count",
    ascending=False
)

print(team_report)
```

```python
# Sort Report Using Multiple Columns from Highest to Lowest

team_report = team_report.sort_values(
    by=[
        "incident_count",
        "average_cpu_usage"
    ],
    ascending=[
        False,
        False
    ]
)

print(team_report)
```

```python
# Rename Columns for Human-Readable Reports
# Business facing reports look cleaner with spaces and capital letters

team_report_display = team_report.rename(
    columns={
        "team": "Team",
        "incident_count": "Incident Count",
        "average_cpu_usage": "Average CPU Usage",
        "highest_cpu_usage": "Highest CPU Usage",
        "average_memory_usage": "Average Memory Usage",
        "average_resolution_minutes": (
            "Average Resolution Minutes"
        )
    }
)

print(team_report_display)
```

```python
# Reorder Report Columns

team_report_display = team_report_display[
    [
        "Team",
        "Incident Count",
        "Average CPU Usage",
        "Highest CPU Usage",
        "Average Memory Usage",
        "Average Resolution Minutes"
    ]
]

print(team_report_display)
```

```python
# Build an Environment Report

environment_report = (
    df.groupby("environment")
      .agg(
          incident_count=("incident_id", "count"),
          critical_incidents=(
              "priority",
              lambda values: (
                  values == "Critical"
              ).sum()
          ),
          average_cpu_usage=("cpu_usage", "mean"),
          highest_cpu_usage=("cpu_usage", "max"),
          average_resolution_minutes=(
              "resolution_minutes",
              "mean"
          )
      )
      .reset_index()
      .round(2)
)

print(environment_report)
```

```python
# Add a Business Metric
# Suppose management considers incidents taking more than 60 mins to be SLA breaches
# Create a new Boolean column

df["sla_breached"] = (
    df["resolution_minutes"] > 60
)

print(
    df[
        [
            "incident_id",
            "resolution_minutes",
            "sla_breached"
        ]
    ]
)
```

```python
# Summarize SLA Performance

sla_report = (
    df.groupby("team")
      .agg(
          total_incidents=("incident_id", "count"),
          sla_breaches=("sla_breached", "sum"),
          average_resolution_minutes=(
              "resolution_minutes",
              "mean"
          )
      )
      .reset_index()
      .round(2)
)

print(sla_report)
```

```python
# Calculate an SLA Compliance Percentage
# First calculate compliant incidents:

sla_report["sla_compliant_incidents"] = (
    sla_report["total_incidents"]
    - sla_report["sla_breaches"]
)

# Then calculate the percentage:

sla_report["sla_compliance_percentage"] = (
    (
        sla_report["sla_compliant_incidents"]
        / sla_report["total_incidents"]
    )
    * 100
)

# Round the final report:

sla_report = sla_report.round(2)

print(sla_report)
```

```python
# Add a Percentage Symbol for Display

sla_report_display = sla_report.copy()

sla_report_display[
    "sla_compliance_display"
] = (
    sla_report_display[
        "sla_compliance_percentage"
    ]
    .astype(str)
    + "%"
)

print(sla_report_display)
```

```python
# Create a Daily Operations Report
# one row per date
daily_report = (
    df.groupby("incident_date")
      .agg(
          incident_count=("incident_id", "count"),
          critical_incidents=(
              "priority",
              lambda values: (
                  values == "Critical"
              ).sum()
          ),
          average_cpu_usage=("cpu_usage", "mean"),
          highest_cpu_usage=("cpu_usage", "max"),
          average_resolution_minutes=(
              "resolution_minutes",
              "mean"
          ),
          sla_breaches=("sla_breached", "sum")
      )
      .reset_index()
      .round(2)
)

print(daily_report)
            
```

```python
# Create a Status Pivot Report

status_pivot_report = (
    pd.pivot_table(
        data=df,
        index="team",
        columns="status",
        values="incident_id",
        aggfunc="count",
        fill_value=0,
        margins=True,
        margins_name="Total"
    )
    .reset_index()
)

status_pivot_report.columns.name = None

print(status_pivot_report)
```

```python
# Exercise 1 - Create a report named priority report. Group by priority and calculate:
# incident_count, average_cpu, highest_cpu, average_resolution_minutes. Reset the index, round
# to 2 decimal points, and sort incident count from highest to lowest

priority_report = (
    df.groupby("priority")
      .agg(
          incident_count=("incident_id", "count"),
          average_cpu=("cpu_usage", "mean"),
          highest_cpu=("cpu_usage", "max"),
          average_resolution_minutes=(
              "resolution_minutes",
              "mean"
          )
      )
      .reset_index()
      .round(2)
)

priority_report = priority_report.sort_values(
    by="incident_count",
    ascending=False
)

# I can include the sorting inside of the method chain (after reset_index and round()) as:

# .sort_values(
#     by="incident_count",
#     ascending=False
# )
# )


print(priority_report)
    
```

```python
#  Exercise 2 - Create a report named server_report. Group by server and calculate number of incidents,
# average cpu usage, maximum memory usage, average resolution time. Sort by average CPU usage from
# highest to lowest

server_report = (
    df.groupby("server")
      .agg(
          incident_count=("incident_id", "count"),
          average_cpu_usage=("cpu_usage", "mean"),
          maximum_memory_usage=("memory_usage", "max"),
          average_resolution_minutes=(
              "resolution_minutes",
              "mean"
          )
      )
      .reset_index()
      .round(2)
)

server_report = server_report.sort_values(
    by="average_cpu_usage",
    ascending=False
)

print(server_report)
          
```

```python
# Exercise 3 - Create a report named environment_sla_report. Group by environment and calculate
# total incidents, total SLA breaches, average resolution time. Then create sla_compliant_incidents and
# sla_compliance percentage. Round the result to 2 decimal places

environment_sla_report = (
    df.groupby("environment")
      .agg(
          total_incidents=("incident_id", "count"),
          sla_breaches=("sla_breached", "sum"),
          average_resolution_time=(
              "resolution_minutes",
              "mean"
          )
      )
      .reset_index() 
)

environment_sla_report["sla_compliant_incidents"] = (
    environment_sla_report["total_incidents"]
    - environment_sla_report["sla_breaches"]
)

environment_sla_report["sla_compliance_percentage"] = (
    (
        environment_sla_report["sla_compliant_incidents"]
        / environment_sla_report["total_incidents"]
    )
    * 100
)

environment_sla_report = environment_sla_report.round(2)

print(environment_sla_report)
```

```python
# Exercise 4 - Create a pivot table named priority_status_pivot. Use Rows:priority; Columns:status; Values:incident_id;
# Calculation:count; fill missing values with 0; add totals; name totals "Total"; reset the index and remove
# the columns index name

priority_status_pivot = (
    pd.pivot_table(
        data=df,
        index="priority",
        columns="status",
        values="incident_id",
        aggfunc="count",
        fill_value=0,
        margins=True,
        margins_name="Total"
    )
    .reset_index()
)

priority_status_pivot.columns.name = None

print(priority_status_pivot)
```

```python
import pandas as pd

server_data = {
    "server": [
        "WEB01",
        "WEB02",
        "DB01",
        "APP01",
        "WEB03",
        "DB02",
        "APP02",
        "WEB04"
    ],
    "status": [
        "Up",
        "Down",
        "Up",
        "Down",
        "Up",
        "Down",
        "Up",
        "Down"
    ],
    "cpu_usage": [
        35,
        92,
        68,
        88,
        27,
        76,
        54,
        81
    ],
    "memory_usage": [
        48,
        95,
        72,
        84,
        39,
        80,
        61,
        86
    ],
    "environment": [
        "Production",
        "Production",
        "Development",
        "Production",
        "Development",
        "Development",
        "Production",
        "Production"
    ],
    "team": [
        "Web",
        "Web",
        "Database",
        "Application",
        "Web",
        "Database",
        "Application",
        "Web"
    ]
}

df = pd.DataFrame(server_data)
print(df)
```

```python
# Counting Servers by Status

server_count_by_status = df.groupby("status").size()

print(server_count_by_status)
```

```python
# Grouping By Multiple Columns

environment_status_groups = (
    df.groupby(["environment", "status"]).size()
)

print(environment_status_groups)
```

```python
# Average CPU by Environment and Status
# Combine multiple grouping columns with an aggregation

average_cpu_groups = (
    df.groupby(["environment", "status"])["cpu_usage"].mean()
)

print(average_cpu_groups)
```

```python
average_cpu_by_status = df.groupby("status")["cpu_usage"].mean()

print(type(average_cpu_by_status))
```

```python
# reset_index

average_cpu_by_status = (
    df.groupby("status")["cpu_usage"].mean().reset_index()
)

print(average_cpu_by_status)
```

```python
print(type(average_cpu_by_status))
```

```python
# Renaming the Calculated Column from cpu_usage -> average_cpu_usage

average_cpu_by_status = (
    df.groupby("status")["cpu_usage"].mean().reset_index().rename(columns={"cpu_usage": "average_cpu_usage"})
)

print(average_cpu_by_status)
```

```python
# Grouping More Than One Numeric Column
# We want the average of both CPU and memory usage by environment

resource_summary = (
    df.groupby("environment")[["cpu_usage", "memory_usage"]].mean().reset_index()
)

print(resource_summary)
```

```python
# Method Chaining - when code is written across several lines
# Sort the Grouped Results - sorts the rows using the average CPU column, from highest to lowest

environment_cpu_report = (
    df.groupby("environment")["cpu_usage"]
        .mean()
        .reset_index()
        .rename(columns={"cpu_usage": "average_cpu_usage"})
        .sort_values(by="average_cpu_usage", ascending=False)
)

print(environment_cpu_report)
```

```python
# Filter Before Grouping

down_servers = df[df["status"] == "Down"]

down_server_summary = (
    down_servers.groupby("environment")["cpu_usage"]
                .mean()
                .reset_index()
)

print(down_server_summary)
```

```python
# Perform Filter and Grouping in One Statement

down_server_summary = (
    df[df["status"] == "Down"]
        .groupby("environment")["cpu_usage"]
        .mean()
        .reset_index()
)

print(down_server_summary)
```

```python
# Real DE Example
# Which support team has the highest average CPU usage?

team_cpu_report = (
    df.groupby("team")["cpu_usage"]
      .mean()
      .reset_index()
      .rename(columns={"cpu_usage": "average_cpu_usage"})
      .sort_values(by="average_cpu_usage", ascending=False)
)

print(team_cpu_report)
```

```python
# Calculate the average memory usage for each status

memory_by_status = (
    df.groupby("status")["memory_usage"]
      .mean()
      .reset_index()
      .rename(columns={"memory_usage": "average_memory_usage"})
)

print(memory_by_status)
```

```python
# Count the number of servers assigned to each team

df.groupby("team")["server"].count()
```

```python
# Calculate the highest CPU usage for each environment
df.groupby("environment")["cpu_usage"].max()
```

```python
# Group the data by both team and status, and count the number of rows in each group

number_of_rows = (
    df.groupby(["team", "status"])
      .size()
      .reset_index(name="server_count")
)

print(number_of_rows)
```

```python
# Clean DataFrame Showing the Average CPU usage by team

average_cpu_team = (
    df.groupby("team")["cpu_usage"]
      .mean()
      .reset_index()
      .rename(columns={"cpu_usage": "average_cpu_usage"})
      .sort_values(by="average_cpu_usage", ascending=False)
)

print(average_cpu_team)
```

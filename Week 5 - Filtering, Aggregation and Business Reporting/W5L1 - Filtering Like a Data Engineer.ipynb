Filtering is one of the most frequently used skills in data engineering.  
Before you can summarize or analyze data, you must always need to isolate the records that matter

```python
import pandas as pd

servers = {
    "server": ["WEB01", "WEB02", "DB01", "APP01", "WEB03"],
    "status": ["Up", "Down", "Up", "Down", "Up"],
    "cpu": [35, 92, 68, 88, 27],
    "environment": ["Production", "Production", "Development", "Production", "Development"]
}

df = pd.DataFrame(servers)

print(df)
```

```python
# Concept 1: Filter with One Condition
# Show only down servers

down_servers = df[df["status"] == "Down"]

print(down_servers)
```

```python
# Concept 2: Filter Numeric Data
# Show servers with CPU usage greater than 80%

high_cpu = df[df["cpu"] > 80]

print(high_cpu)
```

```python
# Concept 3: Multiple Conditions Using AND
# Suppose you only want production servers that are down

critical = df[
    (df["status"] == "Down") &
    (df["environment"] == "Production")
    ]

print(critical)
```

```python
# Concept 3: Multiple Conditions Using OR
# Suppose you only want production servers that are down

critical = df[
    (df["cpu"] > 80) |
    (df["status"] == "Down")
     ]

print(critical)
```

```python
# Concept 4: Using isin()
# Find only these servers:

important = df[df["server"].isin(["WEB01", "DB01"])]

print(important)
```

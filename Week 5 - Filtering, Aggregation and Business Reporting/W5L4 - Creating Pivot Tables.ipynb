```python
import pandas as pd

server_data = {
    "server": [
        "WEB01",
        "WEB02",
        "WEB03",
        "WEB04",
        "DB01",
        "DB02",
        "APP01",
        "APP02",
        "APP03",
        "APP04"
    ],
    "environment": [
        "Production",
        "Production",
        "Development",
        "Development",
        "Production",
        "Development",
        "Production",
        "Development",
        "Production",
        "Development"
    ],
    "status": [
        "Up",
        "Down",
        "Up",
        "Down",
        "Up",
        "Down",
        "Down",
        "Up",
        "Up",
        "Down"
    ],
    "team": [
        "Web",
        "Web",
        "Web",
        "Web",
        "Development",
        "Development",
        "Application",
        "Application",
        "Application",
        "Application"
    ],
    "cpu_usage": [
        35,
        91,
        42,
        76,
        67,
        82,
        94,
        51,
        73,
        88
    ],
    "memory_usage": [
        45,
        93,
        58,
        79,
        72,
        80,
        90,
        60,
        75,
        86
    ],
    "incident_count": [
        1,
        5,
        0,
        3,
        1,
        4,
        6,
        1,
        2,
        5
    ]
}

df = pd.DataFrame(server_data)

print(df)
```

```python
cpu_pivot = pd.pivot_table(
    data=df,
    index="environment",
    columns="status",
    values="cpu_usage",
    aggfunc="mean"
)

print(cpu_pivot)
```

```python
# Average memory usage by team and environment

memory_pivot = pd.pivot_table(
    data=df,
    index="team",
    columns="environment",
    values="memory_usage",
    aggfunc="mean"
)

print(memory_pivot)
```

```python
# Using Other Aggregation Functions
# Maximum CPU Usage by team and environment

max_cpu_pivot = pd.pivot_table(
    data=df,
    index="team",
    columns="environment",
    values="cpu_usage",
    aggfunc="max"
)

print(max_cpu_usage)
```

```python
# Minimum CPU Usage by team and environment

min_cpu_pivot = pd.pivot_table(
    data=df,
    index="team",
    columns="environment",
    values="cpu_usage",
    aggfunc="min"
)

print(min_cpu_usage)
```

```python
# Sum - What is the total number of incidents for each team in each environment?

incident_total_pivot = pd.pivot_table(
    data=df,
    index="team",
    columns="environment",
    values="incident_count",
    aggfunc="sum"
)

print(incident_total_pivot)
```

```python
# Count the number of non-missing server values for each team and status

server_count_pivot = pd.pivot_table(
    data=df,
    index="team",
    columns="status",
    values="server",
    aggfunc="count"
)

print(server_count_pivot)
```

```python
# Using aggfunc="size" to Count Rows

server_count_pivot = pd.pivot_table(
    data=df,
    index="team",
    columns="status",
    aggfunc="size"
)

print(server_count_pivot)
```

```python
# Multiple Aggregation Functions / MultiIndex

multiple_calculations = pd.pivot_table(
    data=df,
    index="environment",
    columns="status",
    values="cpu_usage",
    aggfunc=["mean", "max", "min"]
)

print(multiple_calculations)
```

```python
# Multiple Value Columns

resource_pivot = pd.pivot_table(
    data=df,
    index="environment",
    columns="status",
    values=["cpu_usage", "memory_usage"],
    aggfunc="mean"
)

print(resource_pivot)
```

```python
# Filling Missing Values

import pandas as pd

sample_data = {
    "team": [
        "Web",
        "Web",
        "Database",
        "Application"
    ],
    "status": [
        "Up",
        "Down",
        "Up",
        "Down"
    ],
    "incident_count": [
        1,
        5,
        2,
        6
    ]
}

sample_df = pd.DataFrame(sample_data)

print(sample_df)
```

```python
# NaN

sample_pivot = pd.pivot_table(
    data=sample_df,
    index="team",
    columns="status",
    values="incident_count",
    aggfunc="sum"
)

print(sample_pivot)
```

```python
# Replace missing values with fill_value=0

sample_pivot = pd.pivot_table(
    data=sample_df,
    index="team",
    columns="status",
    values="incident_count",
    aggfunc="sum",
    fill_value=0
)

print(sample_pivot)
```

```python
# Adding Totals with margins=True

incident_pivot = pd.pivot_table(
    data=df,
    index="team",
    columns="environment",
    values="incident_count",
    aggfunc="sum",
    margins=True
)

print(incident_pivot)
```

```python
# Renaming the Totals Lable - rename 'All'

incident_pivot = pd.pivot_table(
    data=df,
    index="team",
    columns="environment",
    values="incident_count",
    aggfunc="sum",
    margins=True,
    margins_name="Total"
)

print(incident_pivot)
```

```python
# Resetting Pivot Table Index

cpu_pivot = cpu_pivot.reset_index()

print(cpu_pivot)
```

```python
# Removing the Column Index Name

cpu_pivot = pd.pivot_table(
    data=df,
    index="environment",
    columns="status",
    values="cpu_usage",
    aggfunc="mean"
).reset_index()

cpu_pivot.columns.name = None

print(cpu_pivot)
```

```python
# Rounding Pivot Table Results for Decimal Places

cpu_pivot = pd.pivot_table(
    data=df,
    index="environment",
    columns="status",
    values="cpu_usage",
    aggfunc="mean"
).round(2)

print(cpu_pivot)
```

```python
# Complete Report-Ready Pivot Table

incident_report = (
    pd.pivot_table(
        data=df,
        index="team",
        columns="environment",
        values="incident_count",
        aggfunc="sum",
        fill_value=0,
        margins=True,
        margins_name="Total"
    )
    .reset_index()
)

incident_report.columns.name = None

print(incident_report)
```

```python
# Exporting a Pivot Table

incident_report.to_csv(
    "incident_summary_by_team.csv",
    index=False
)
```

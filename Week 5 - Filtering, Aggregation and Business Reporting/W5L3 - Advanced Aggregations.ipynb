```python
# Create Our Dataset

import pandas as pd

server_data = {
    "server": [
        "WEB01", "WEB02", "WEB03",
        "DB01", "DB02",
        "APP01", "APP02", "APP03"
    ],
    "environment": [
        "Production", "Production", "Development",
        "Development", "Production",
        "Production", "Development", "Production"
    ],
    "status": [
        "Up", "Down", "Up",
        "Down", "Up",
        "Down", "Up", "Down"
    ],
    "team":[
        "Web", "Web", "Web",
        "Database", "Database",
        "Application", "Application", "Application"
    ],
    "cpu_usage": [
        35, 91, 42,
        82, 67, 
        94, 51, 87
    ],
    "memory_usage":[
        45, 93, 58,
        80, 72, 
        90, 60, 85
    ]
}

df = pd.DataFrame(server_data)

print(df)
```

```python
# With agg() we can ask for different types of aggregation calculations at once

cpu_summary = (
    df.groupby("environment")["cpu_usage"]
      .agg(["mean", "max", "min"])
)

print(cpu_summary)
```

```python
# Adding Count

cpu_summary = (
    df.groupby("environment")["cpu_usage"]
      .agg(["count", "mean", "max", "min"])
)
print(cpu_summary)
```

```python
# Named Aggregations
# rename agg columns

cpu_summary = (
    df.groupby("environment")
      .agg(
          average_cpu=("cpu_usage", "mean"),
          highest_cpu=("cpu_usage", "max"),
          lowest_cpu=("cpu_usage", "min")
      )
)

print(cpu_summary)
```

```python
# Multiple Columns

summary = (
    df.groupby("environment")
      .agg(
          average_cpu=("cpu_usage", "mean"),
          average_memory=("memory_usage", "mean"),
          highest_cpu=("cpu_usage", "max"),
          highest_memory=("memory_usage", "max")
      )
)

print(summary)
          
```

```python
# transform()

df["environment_average_cpu"] = (
    df.groupby("environment")["cpu_usage"]
      .transform("mean")
)

print(df)
```

```python
# Exercise 1
# Create a report showing for each team: count, average, maximum, minimum CPU

report = (
    df.groupby("team")["cpu_usage"]
      .agg(["count", "mean", "max", "min"])
)

print(report)
```

```python
# Exercise 2
# Create a named aggregatio report by status with columns:
# average_memory, highest_memory, lowest_memory

report_by_status = (
    df.groupby("status")
      .agg(
          average_memory=("memory_usage", "mean"),
          highest_memory=("memory_usage", "max"),
          lowest_memory=("memory_usage", "min")
      )
)

print(report_by_status)
```

```python
# Create Time Categories - morning, afternoon, evening

def classify_time_of_day(hour):
    if hour < 12:
        return "Morning"
    elif hour < 17:
        return "Afternoon"
    else:
        return "Evening"

df["time_of_day"] = (
    df["incident_time"]
        .dt.hour
        .apply(classify_time_of_day)
)

print(
    df[
        [
            "incident_time",
            "incident_hour",
            "time_of_day"
        ]
    ] 
)
```

```python
# Find incidents that took longer than one hour

long_incidents = df[
    df["resolution_minutes"] > 60
    ]

print(
    long_incidents[
        [
            "incident_id",
            "server",
            "resolution_minutes"
        ]
    ]
)
```

```python
# Complete Time-Series Workflow

import pandas as pd

incident_data = {
    "incident_id": ["INC001", "INC002", "INC003", "INC004"],
    "server": ["WEB01", "DB01", "APP01", "WEB02"],
    "cpu_usage": [92, 76, 95, 81],
    "incident_time": [
        "2026-07-01 08:15:00",
        "2026-07-01 10:30:00",
        "2026-07-02 14:45:00",
        "2026-07-02 16:40:00"
    ],
    "resolved_time": [
        "2026-07-01 09:00:00",
        "2026-07-01 10:50:00",
        "2026-07-02 16:15:00",
        "2026-07-02 16:40:00"
    ]
    
}

incidents_df = pd.DataFrame(incident_data)

incidents_df["incident_time"] = pd.to_datetime(
    incidents_df["incident_time"]
)

incidents_df["resolved_time"] = pd.to_datetime(
    incidents_df["resolved_time"]
)

incidents_df["incident_date"] = (
    incidents_df["incident_time"].dt.date
)

incidents_df["incident_hour"] = (
    incidents_df["incident_time"].dt.hour
)

incidents_df["resolution_minutes"] = (
    (
        incidents_df["resolved_time"]
        - incidents_df["incident_time"]
    )
    .dt.total_seconds()
    / 60
)

daily_report = (
    incidents_df
    .groupby("incident_date")
    .agg(
        incident_count=("incident_id", "count"),
        average_cpu=("cpu_usage", "mean"),
        average_resolution_minutes=("resolution_minutes", "mean")
    )
    .reset_index()
    .round(2)
)

print(incidents_df)
print(daily_report)
    
```

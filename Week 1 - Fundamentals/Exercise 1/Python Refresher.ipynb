```python
# Step 3: Loop Through the Alerts - we need to examine each item
for alert in alerts:
    print(alert)

# What the Loop Says: For every alert inside alerts, do something
```

### **What is an iteration?**
An iteration is simply one pass through a loop.

When Python sees:
"for alert in alerts" it says: Take one item from the 'alerts' list at a time and store it in the variable 'alert'

Given the 'alerts' list, Python will go through the list one item at a time.

### **Before the Loop Starts**
We create an empty dictionary: counts = {}, where nothing is stored yet.

**Iteration 1**  
Python grabs the 1st item: alert = "Critical"  
Now the loop body runs: if alert in counts:  
This becomes: if "Critical" in counts:  

**Question**  
Is there already a key called "Critical" in the dictionary?  
Current dictionary: {} Answer is "False"  
Because the dictionary is empty, Python goes to: else: counts[alert]=1  
Substitute the value: counts["Critical"] = 1; Dictionary now becomes {"Critical":1}  
End of Iteration 1  

**Iteration 2**  
Python grabs the next item: alert = "High"  
Dictionary currently: {"Critical":1}  
Check: if "High" in counts:  

**Question**  
Does the dictionary contain a key named "High"?  
Current keys: "Critical", the answer is "False"  
Python executes: counts["High"] = 1  
Dictionary becomes: {"Critical":1, "High":1}  
End of Iteration 2

**Iteration 3**  
Python grabs: alert = "Critical"  
Dictionary currently: {"Critical":1, "High":1}  
Check: if "Critical" in counts:  

**Question**  
Is there already a key called "Critical"? "True"  
So Python executes: counts["Critical"] += 1  
Let's expand - Current value: counts["Critical"] is 1; counts["Critical"] += 1  
Means: counts["Critical"] = counts["Critical"] + 1; becomes: counts["Critical"] = 1 + 1  
Which becomes: counts["Critical"] = 2  
Dictionary now: {"Critical": 2, "High": 1}  
End of Iteration 3  

**Iteration 4**  
Python grabs: alert = "Medium"  
Dictionary: {"Critical": 2, "High": 1}  
Check: if "Medium" in counts: "False"  
So: counts["Medium"] = 1  
Dictionary becomes: {"Critical": 2, "High": 1, "Medium": 1}  
End of Iteration 4  

**Iteration 5**  
Python grabs: alert = "High"  
Dictionary: {"Critical": 2, "High": 1, "Medium": 1}  
Check: if "High" in counts: "True"  
Execute: counts["High"] += 1; Current value: 1  
New value: 2; Dictionary becomes: {"Critical": 2, "High": 2, "Medium": 1}  
End of Iteration 5  

**Iteration 6**  
Python grabs: alert = "Critical"  
Dictionary: {"Critical": 2, "High": 2, "Medium": 1}  
Check: if "Critical" in counts: "True"  
Execute: counts["Critical"] += 1; Current value: 2  
New value: 3; Dictionary becomes: {"Critical": 3, "High": 2, "Medium": 1}  
End of Iteration 6

### **Loop Finished**
Python has reached the end of the list.  
There are no more alerts to process.  
Now: Python return counts  

```python
# Step 4: Count Occurrences
# For each alert: If we've seen it before, add 1; If we haven't seen it before, start it at 1
# Loop explained step by step in above cell

# Final Solution:

def count_alerts(alerts):
    counts = {}

    for alert in alerts:
        formatted_alert = alert.title()
        
        if formatted_alert in counts:
            counts[alert.title()] += 1

        else:
            counts[alert.title()] = 1

    return counts
```

```python
# Testing:

result = count_alerts(alerts)

print(result)
```

```python
# Step 1: Import JSON
import json
```

```python
incidents = [
    {"server": "WEB01", "status": "Down"},
    {"server": "WEB02", "status": "Up"},
    {"server": "WEB03", "status": "Down"}
]

# incidents = list
# each item = dictionary
```

```python
with open("incidents.json", "w") as file:
    json.dump(incidents, file, indent=4)
```

```python
with open("incidents.json", "r") as file:
    loaded_incidents = json.load(file)

print(loaded_incidents)
print(type(loaded_incidents))
```

```python
incidents = [
    {"id": 1001, "server": "WEB01", "status": "Up", "cpu": 35},
    {"id": 1002, "server": "WEB02", "status": "Down", "cpu": 92},
    {"id": 1003, "server": "DB01", "status": "Up", "cpu": 78},
    {"id": 1004, "server": "WEB03", "status": "Down", "cpu": 99},
]
```

```python
# Task 1: Display every server
# Write a loop that prints: WEB01, WEB02, DB01, WEB03

for incident in incidents:
    print(incident["server"])
```

```python
# Task 2: Display only down servers
# Write a loop where the expected output is: WEB02 is Down; WEB03 is Down

for incident in incidents:
    if incident["status"] == "Down":
        print(incident["server"], "is Down")
```

```python
# Task 3: Find High CPU Servers
# Print servers where CPU usage is greater than 80

for incident in incidents:
    if incident["cpu"] > 80:
        print(incident["cpu"])
```

```python
# Task 4: Count Down Servers
# Create a variable named: down_count = 0
# Loop through the incidents and count how many servers have: status == "Down"
# Expected result: 2

down_count = 0

for incident in incidents:
    if incident["status"] == "Down":
        down_count = down_count + 1 # Reads as: take the current value of down_count, add 1, and store
                                    # the result back into down_count

print(down_count)
```

When you're counting something, for example: counter = 0; it usually goes before the loop begins

```python
# Task 5: Create a Function
# Create a function that returns a list of server names that are down
# Expected result: ["WEB02", "WEB03"]

def get_down_servers(data):
    down_servers = []

    for incident in data:
        if incident["status"] == "Down":
            down_servers.append(incident["server"])

    return down_servers

# This portion of the code builds the list of results that we want printed out
```

```python
result = get_down_servers(incidents)
print(result)

# This portion prints out the actual results that we want
```

```python
# Step 1 - Create a CSV

import pandas as pd

data = {
    "server": ["WEB01", "WEB02", "WEB03", "WEB04", "WEB05", "WEB06", "WEB07", "WEB08"],
    "status": ["Up", "Down", "Up", "Down", "Up", "Up", "Down", "Up"],
    "cpu": [35, 92, 45, 97, 88, 15, 99, 65],
    "team": ["Web", "Web", "Database", "Database", "Web", "Application", "Application", "Database"]
}

df = pd.DataFrame(data)

df.to_csv("server_health.csv", index=False)
```

```python
# Step 2 - Read the File

import pandas as pd

df = pd.read_csv("server_health.csv")
```

```python
# Step 3 - Explore the Data

print(df.head())
```

```python
# Step 3 - Explore the Data

print(df.shape)
```

```python
# Step 3 - Explore the Data

print(df.columns)
```

```python
# Step 4 - Find Down Servers

down_servers = df[df["status"] == "Down"]

print(down_servers)
```

```python
# Step 5 - Count Down Servers

print(len(down_servers))
```

```python
# Step 6 - Create CPU Warning Column
# adds CPU Warning Column to existing data

df["cpu_warning"] = df["cpu"] > 90
print(df)
```

```python
# Step 7 - Sort by CPU Usage

sorted_df = df.sort_values("cpu", ascending=False)
print(sorted_df)
```

```python
# Step 8 - Count Status Values

print(df["status"].value_counts())
```

```python
# Step 9 - Group by Team
# Total CPU by Team

print(df.groupby("team")["cpu"].sum())
```

```python
# Step 9 - Group by Team
# Average CPU by Team

print(df.groupby("team")["cpu"].mean())
```

```python
# Step 10 - Team Summary Report

print(df.groupby("team")["cpu"].agg(["count", "sum", "mean", "max"]))
```

```python
import pandas as pd

data = {
    "server": ["WEB01", "WEB02", "WEB03", "WEB04", "WEB05"],
    "status": ["Up", "Down", "Up", "Down", "Up"],
    "cpu": [35, 92, 45, 97, 88]
}

df = pd.DataFrame(data)

df.to_csv("servers.csv", index=False)
```

```python
import os
print(os.listdir())
```

```python
df["cpu_warning"] = df["cpu"] > 90
print(df)
```

```python
df["cpu_status"] = "Normal"
```

```python
df.loc[df["cpu"] > 90, "cpu_status"] = "Critical"
```

```python
print(df)
```

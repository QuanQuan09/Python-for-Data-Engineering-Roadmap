```python
# value_counts()

import pandas as pd

data = {
    "status": ["Up", "Down", "Up", "Down", "Up"]
}

df = pd.DataFrame(data)
```

```python
df["status"].value_counts()
```

```python
# Check for Missing Data

import pandas as pd

data = {
    "server": ["WEB01", "WEB02", "WEB03"],
    "cpu": [35, None, 45]
}

df = pd.DataFrame(data)
```

```python
# Find Missing Values

df.isnull()
```

```python
# Count Missing Values

df.isnull().sum()
```

```python
# Removing Missing Values

df.dropna()
```

```python
# Filling Missing Values
#Instead of deleting

df["cpu"] = df["cpu"].fillna(0)
```

```python
df.isnull()
```

```python
# Removing Duplicate Records

data = {
    "server": ["WEB01", "WEB02", "WEB02", "WEB03"]
}

df = pd.DataFrame(data)
```

```python
df.drop_duplicates()
```

```python
df["server"].value_counts()
```

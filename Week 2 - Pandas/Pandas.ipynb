# Data Engineer Tip

Whenever working with a new dataset, get into the habit of running these commands first:  
df = pd.read_csv("enter csv.csv")  
print(df.head()) - shows the first 5 rows  
print(df.shape()) - shows the structure of the table   
print(df.columns) - shows all column names  
print(df.info()) - shows number of rows, columns, data types, and missing values

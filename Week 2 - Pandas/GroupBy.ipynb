```python
import pandas as pd

data = {
    "team": ["Web", "Web", "DB", "DB", "Web"],
    "status": ["Up", "Down", "Up", "Down", "Up"],
    "cpu": [35, 92, 45, 97, 88]
}

df = pd.DataFrame(data)
```

```python
df.groupby("team")["cpu"].mean()
```

```python
df.groupby("team")["cpu"].count()
```

```python
df.groupby("team")["cpu"].max()
```

```python
df.groupby("team")["cpu"].sum()
```

```python
team_cpu = df.groupby("team")["cpu"].sum()
print(team_cpu)
```

```python
print(df.groupby("team")["cpu"].sum())
```

```python
# multiple calculations at once

df.groupby("team")["cpu"].agg(["count", "sum", "mean", "max"])
```
